```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

df1['date'] = pd.to_datetime(df1['month'], format='%Y-%b')
df1['sector'] = df1['sector'].str.replace('sector ', '', regex=False).astype(int)
df1 = df1.sort_values(by=['date'])

sectores = df1['sector'].unique()
n = len(sectores)
fig, axes = plt.subplots(nrows=(n // 5) + (n % 5 > 0), ncols=5, figsize=(20, 4 * ((n // 5) + 1)), sharey=True)
axes = axes.flatten()

for i, sector in enumerate(sectores):
    sector_df = df1[df1['sector'] == sector]
    axes[i].plot(sector_df['date'], sector_df['amount_new_house_transactions'], marker='o')
    axes[i].set_title(sector)
    axes[i].grid(True, linestyle='--', alpha=0.6)

    # Forzar formato de fechas en X
    axes[i].xaxis.set_major_formatter(mdates.DateFormatter("%Y-%m"))
    axes[i].tick_params(axis='x', rotation=45)

for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

fig.suptitle('Evolución de amount_new_house_transactions por sector', fontsize=16)
plt.tight_layout()
plt.show()
```
