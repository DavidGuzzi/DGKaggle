```python
import os
import zipfile

DATA_DIR = "../data"
COMPETITION = "playground-series-s6e6"

os.makedirs(DATA_DIR, exist_ok=True)

!kaggle competitions download -c {COMPETITION} -p {DATA_DIR}

# Descomprimir el archivo descargado
zip_path = os.path.join(DATA_DIR, f"{COMPETITION}.zip")
if os.path.exists(zip_path):
    with zipfile.ZipFile(zip_path, "r") as z:
        z.extractall(DATA_DIR)
    os.remove(zip_path)
    print("Datos extraidos y zip eliminado.")
else:
    print("No se encontro el zip. Verificar credenciales y nombre de competencia.")
```

```python
# Verificar archivos descargados
for f in os.listdir(DATA_DIR):
    size_mb = os.path.getsize(os.path.join(DATA_DIR, f)) / (1024 * 1024)
    print(f"{f:40s} {size_mb:.2f} MB")
```
